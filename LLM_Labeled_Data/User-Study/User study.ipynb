```python
import pandas as pd
import glob
import os

# Get all CSV files in the current folder (adjust the pattern if needed)
csv_files = glob.glob("*.csv")

dfs = []
for f in csv_files:
    print("Reading:", f)
    # Read user_id as *string* so pandas never converts it to scientific notation
    df = pd.read_csv(f, dtype={"user_id": str})
    dfs.append(df)

# Concatenate all dataframes
merged = pd.concat(dfs, ignore_index=True)

# Save merged file
output_file = "merged_all.csv"
merged.to_csv(output_file, index=False)

print("Saved merged file as:", output_file)
merged
```
